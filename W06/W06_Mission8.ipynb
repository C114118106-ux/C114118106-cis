## Mission 5：條件判斷與自訂函式

**Knowledge Point**：`def`、參數、`return`、`if / elif / else`、縮排

函式把一段可重複使用的規則取名。`score` 是輸入參數；`return` 傳回結果。Python 用縮排表示哪些指令屬於函式或條件區塊。

**執行前預測**：輸入 82、70、55 時各會回傳什麼文字？

````python
# TODO：補上三種分數範圍的回傳文字
def describe_score(score):
    if score >= 80:
        return "表現良好"
    elif score >= 60:
        return "達到基本標準"
    else:
        return "需要加強"

assert describe_score(82) == "表現良好", "檢查 80 分以上的分支。"
assert describe_score(70) == "達到基本標準", "檢查 60–79 分的分支。"
assert describe_score(55) == "需要加強", "檢查 60 分以下的分支。"
print(describe_score(82), describe_score(70), describe_score(55))
````

## Mission 7：建立模擬的大學生學習資料

**Knowledge Point**：亂數、常態分布、`clip()`、DataFrame、列與欄

每列代表一位**模擬**學生，每欄是一項變數。`rng.normal(平均, 標準差, 筆數)` 讓數值有差異；這裡不要求理解常態分布的數學原理。`clip()` 將數值限制在指定範圍。`df` 是資料表變數名稱。

**執行前預測**：資料會有幾列、幾欄？`df.head()` 會顯示幾列？

````python
# 先使用上一個 Mission 的 student_count、departments、rng
n = student_count
student_ids = np.arange(1, n + 1)  # TODO：用 np.arange 建立 1 到 60 的編號
chosen_departments = rng.choice(departments, size=n)
sleep_values = rng.normal(6.7, 1.1, size=n).round(1)

raw_screen_values = rng.normal(5.0, 1.5, size=n)
screen_values = raw_screen_values.clip(1, 10).round(1)  # TODO：將 raw_screen_values 限制在 1 到 10，再四捨五入

raw_study_values = rng.normal(2.5, 1.0, size=n)
study_values = raw_study_values.clip(0.5, 6).round(1)  # TODO：將 raw_study_values 限制在 0.5 到 6，再四捨五入

attendance_values = rng.normal(88, 8, size=n)
attendance_values = attendance_values.clip(60, 100).round().astype(int)

assert student_ids is not None, "請先建立學生編號。"
assert screen_values is not None, "請先處理螢幕時間範圍。"
assert study_values is not None, "請先處理讀書時間範圍。"
df = pd.DataFrame({
    "student_id": student_ids,
    "department": chosen_departments,
    "sleep_hours": sleep_values,
    "screen_time_hours": screen_values,
    "study_hours": study_values,
    "attendance_rate": attendance_values,
})
display(df.head())
assert df.shape == (60, 6), "請檢查資料列數與欄數。"
````

## Mission 8：建立小考成績欄位

**Knowledge Point**：欄位運算、隨機誤差、多欄選取、分數限制

沿用 Mission 4 的人為公式，另外加入平均數 0、標準差 5 的隨機誤差。`df["欄名"]` 選一欄；`df[["欄1", "欄2"]]` 選多欄。

**執行前預測**：新欄位加入後，資料表共有幾欄？所有分數應落在哪個範圍？

````python
# 沿用 Mission 4 的人為公式，逐項計算
study_part = 3 * df["study_hours"]
sleep_part = 2 * df["sleep_hours"]       # TODO：2 × 睡眠時數欄位
screen_part = -1.2 * df["screen_time_hours"]      # TODO：-1.2 × 螢幕時間欄位
attendance_part = 0.15 * df["attendance_rate"]
random_error = rng.normal(0, 5, size=n)

assert sleep_part is not None and screen_part is not None, "請先補完欄位運算。"
raw_scores = 50 + study_part
raw_scores = raw_scores + sleep_part
raw_scores = raw_scores + screen_part
raw_scores = raw_scores + attendance_part
raw_scores = raw_scores + random_error

# TODO：限制在 0 到 100，四捨五入，再轉成整數
df["quiz_score"] = raw_scores.clip(0, 100).round().astype(int)
assert df["quiz_score"].notna().all(), "請先建立 quiz_score。"
assert df["quiz_score"].between(0, 100).all(), "分數應在 0 到 100。"
display(df[["student_id", "study_hours", "quiz_score"]].head())
display(df)
````
