### Markdown Cell

我們為一位假想學生手算「小考成績」。請勿一次寫成一行，而是分步驟計算每項影響，最後相加。這樣容易除錯、也易閱讀。

⚠️ 提醒：這是人為設計的模擬公式，不代表真實因果關係。

```python
# Mission 4：逐步計算模擬成績
study_hours = 2.5
sleep_hours = 7.0
screen_time_hours = 4.0
attendance_rate = 90

# 各項影響分開計算
base_score = 50
study_effect = 3 * study_hours
sleep_effect = 2 * sleep_hours
screen_effect = -1.2 * screen_time_hours
attendance_effect = 0.15 * attendance_rate

# 合計
quiz_score = base_score + study_effect + sleep_effect + screen_effect + attendance_effect

print("基礎分：", base_score)
print("讀書加分：", study_effect)
print("睡眠加分：", sleep_effect)
print("螢幕扣分：", screen_effect)
print("出席加分：", attendance_effect)
print("-------------------")
print("模擬小考成績：", quiz_score)
```

## Mission 5：條件判斷與自訂函式
### Knowledge Point

### Markdown Cell

將「分數轉文字描述」封裝成函式 describe_score()。以後只要呼叫它，就能重複使用邏輯。

```python
# Mission 5：自訂函式與條件判斷
def describe_score(score):
    """根據分數回傳文字描述"""
    if score >= 80:
        return "表現良好"
    elif score >= 60:
        return "達到基本標準"
    else:
        return "需要加強"

# 測試函式
print("分數 85：", describe_score(85))
print("分數 70：", describe_score(70))
print("分數 50：", describe_score(50))
print("分數 60：", describe_score(60))  # 邊界測試
```

資料科學常用 numpy（數值計算）與 pandas（資料表操作）。np.random.default_rng(42) 建立一個可重現的亂數產生器——同一個種子，產生相同序列。

```python
# Mission 6：匯入套件與建立亂數產生器
import numpy as np
import pandas as pd

print("numpy 版本：", np.__version__)
print("pandas 版本：", pd.__version__)

# 可重現的亂數產生器
random_seed = 42
rng = np.random.default_rng(random_seed)

# 測試：產生 3 個 0~1 之間的亂數
print("前 3 個亂數：", rng.random(3))
```

## Mission 7：建立模擬的大學生學習資料
### Knowledge Point
pandas.DataFrame：表格型資料結構

rng.normal(mean, std, size)：常態分布亂數

.round(1)：四捨五入到小數點後 1 位

.clip(min, max)：限制數值範圍

astype(int)：轉整數

### Markdown Cell
我們要一次產生 60 位學生 的模擬資料。每個欄位用常態分布產生，再用 clip() 避免不合理數值（如負數睡眠時間）。最後組裝成 DataFrame df。

一列 = 一位學生，一欄 = 一項變數，df 就像 Excel 工作表。

```python
# Mission 7：建立 60 筆模擬學生資料
n = 60  # 學生數

# 1. 學生編號 1~60
student_id = list(range(1, n + 1))

# 2. 科系：從三個科系隨機選擇
departments_list = ["資訊管理", "企業管理", "金融管理"]
department = rng.choice(departments_list, size=n)

# 3. 睡眠時數：常態分布 平均 6.7、標準差 1.1，四捨五入到小數點後 1 位
sleep_hours = rng.normal(loc=6.7, scale=1.1, size=n).round(1)

# 4. 螢幕使用時數：常態分布 平均 5.0、標準差 1.5，限制 1~10 小時
screen_time_hours = rng.normal(loc=5.0, scale=1.5, size=n).round(1)
screen_time_hours = np.clip(screen_time_hours, 1.0, 10.0)

# 5. 讀書時數：常態分布 平均 2.5、標準差 1.0，限制 0.5~6 小時
study_hours = rng.normal(loc=2.5, scale=1.0, size=n).round(1)
study_hours = np.clip(study_hours, 0.5, 6.0)

# 6. 出席率：常態分布 平均 88、標準差 8，限制 60~100，轉整數
attendance_rate = rng.normal(loc=88, scale=8, size=n).round(0)
attendance_rate = np.clip(attendance_rate, 60, 100).astype(int)

# 組裝成 DataFrame
df = pd.DataFrame({
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

# 顯示前 5 筆
print("前 5 筆資料：")
print(df.head())
print("\n資料形狀 (列, 欄)：", df.shape)
```

## Mission 8：建立小考成績欄位
### Knowledge Point
DataFrame 新增欄位：df["new_col"] = ...

向量化運算：整欄直接運算，不需迴圈

多欄選取：df[["col1", "col2"]]

加入隨機誤差模擬真實世界未觀測因素
### Markdown Cell
根據 Mission 4 的公式，為 60 位學生一次計算成績，並加入「隨機誤差」（平均 0、標準差 5）。最後用 clip(0, 100) 限制分數範圍，四捨五入成整數。

```python
# Mission 8：計算小考成績並加入 df
# 各項影響分開計算（向量化運算）
base_score = 50
study_effect = 3 * df["study_hours"]
sleep_effect = 2 * df["sleep_hours"]
screen_effect = -1.2 * df["screen_time_hours"]
attendance_effect = 0.15 * df["attendance_rate"]
random_error = rng.normal(loc=0, scale=5, size=n).round(1)  # 隨機誤差

# 合計
df["quiz_score"] = (
    base_score
    + study_effect
    + sleep_effect
    + screen_effect
    + attendance_effect
    + random_error
).round(0).clip(0, 100).astype(int)

# 顯示關鍵欄位前 5 筆
cols_to_show = [
    "student_id",
    "study_hours",
    "sleep_hours",
    "screen_time_hours",
    "attendance_rate",
    "quiz_score"
]
print(df[cols_to_show].head())
```

## Mission 9：加入缺失值與異常值
### Knowledge Point
.loc[列索引, 欄位名稱]：依標籤定位

np.nan：Not a Number，表示缺失值

DataFrame 索引預設從 0 開始 → student_id = 6 對應索引 5
### Markdown Cell
真實資料常有缺失與異常。我們故意製造：

索引 5、17（即學號 6、18）的 sleep_hours 設為 np.nan

索引 8（學號 9）的 screen_time_hours 設為 15.0（超過合理上限）

```python
# Mission 9：植入缺失值與異常值
import numpy as np  # 確保 np 可用

# 缺失值：索引 5、17 的睡眠時數
df.loc[[5, 17], "sleep_hours"] = np.nan

# 異常值：索引 8 的螢幕使用時間設為 15.0
df.loc[8, "screen_time_hours"] = 15.0

# 驗證：印出這三筆資料
print("索引 5 (學號 6)：")
print(df.loc[5, ["student_id", "sleep_hours", "screen_time_hours"]])
print("\n索引 8 (學號 9)：")
print(df.loc[8, ["student_id", "sleep_hours", "screen_time_hours"]])
print("\n索引 17 (學號 18)：")
print(df.loc[17, ["student_id", "sleep_hours", "screen_time_hours"]])
```
